```python
%%writefile StringComparison.java
import java.util.*;

public class StringComparison {
    interface TicketBuilder { String build(int tickets, int fields); }

    static String withString(int tickets, int fields) {
        String last = "";
        for (int t = 0; t < tickets; t++) {
            String s = "";
            for (int f = 0; f < fields; f++) s += "FIELD" + f + "=" + (t + f) + ";";
            last = s;
        }
        return last;
    }

    static String withBuilder(int tickets, int fields) {
        String last = "";
        for (int t = 0; t < tickets; t++) {
            StringBuilder sb = new StringBuilder();
            for (int f = 0; f < fields; f++) sb.append("FIELD").append(f).append('=').append(t + f).append(';');
            last = sb.toString();
        }
        return last;
    }

    static String withBuffer(int tickets, int fields) {
        String last = "";
        for (int t = 0; t < tickets; t++) {
            StringBuffer sb = new StringBuffer();
            for (int f = 0; f < fields; f++) sb.append("FIELD").append(f).append('=').append(t + f).append(';');
            last = sb.toString();
        }
        return last;
    }

    static long benchmark(TicketBuilder builder, int tickets, int fields) {
        builder.build(Math.min(100, tickets), fields); // warm-up
        long start = System.nanoTime();
        String result = builder.build(tickets, fields);
        long elapsed = System.nanoTime() - start;
        if (result.length() == 0) System.out.println("Empty result");
        return elapsed;
    }

    public static void main(String[] args) {
        int[][] workloads = {{1, 6}, {10000, 50}, {2000, 100}};
        TicketBuilder[] builders = {StringComparison::withString, StringComparison::withBuilder, StringComparison::withBuffer};
        String[] names = {"String +", "StringBuilder", "StringBuffer"};

        for (int[] w : workloads) {
            System.out.println("\nTickets=" + w[0] + ", fields=" + w[1]);
            for (int i = 0; i < builders.length; i++) {
                long total = 0;
                for (int r = 0; r < 5; r++) total += benchmark(builders[i], w[0], w[1]);
                System.out.printf("%s average: %.3f ms%n", names[i], total / 5.0 / 1_000_000);
            }
        }
        Runtime rt = Runtime.getRuntime();
        System.out.println("Approx. used memory: " + (rt.totalMemory() - rt.freeMemory()) / 1024 + " KB");
    }
}
```

```python
%%writefile PlaylistComparison.java
import java.util.*;

public class PlaylistComparison {
    static long time(Runnable task) {
        long start = System.nanoTime();
        task.run();
        return System.nanoTime() - start;
    }

    static void test(int n) {
        System.out.println("\nPlaylist size: " + n);

        List<Integer> a = new ArrayList<>();
        long arrayAdd = time(() -> { for (int i = 0; i < n; i++) a.add(i); });
        long arrayGet = time(() -> { long sum = 0; for (int i = 0; i < n; i++) sum += a.get(i); });
        long arrayIterate = time(() -> { long sum = 0; for (int x : a) sum += x; });
        long arrayFront = time(() -> { for (int i = 0; i < 100; i++) a.add(0, i); });

        List<Integer> l = new LinkedList<>();
        long linkedAdd = time(() -> { for (int i = 0; i < n; i++) l.add(i); });
        long linkedGet = time(() -> { long sum = 0; for (int i = 0; i < n; i++) sum += l.get(i); });
        long linkedIterate = time(() -> { long sum = 0; for (int x : l) sum += x; });
        long linkedFront = time(() -> { for (int i = 0; i < 100; i++) l.add(0, i); });

        Deque<Integer> d = new ArrayDeque<>();
        long dequeEnds = time(() -> {
            for (int i = 0; i < n; i++) d.addLast(i);
            for (int i = 0; i < 100; i++) d.addFirst(i);
            for (int i = 0; i < 100; i++) d.pollFirst();
        });

        System.out.printf("ArrayList: add=%d ms, get-all=%d ms, iterate=%d ms, 100 front inserts=%d ms%n",
                arrayAdd/1_000_000, arrayGet/1_000_000, arrayIterate/1_000_000, arrayFront/1_000_000);
        System.out.printf("LinkedList: add=%d ms, get-all=%d ms, iterate=%d ms, 100 front inserts=%d ms%n",
                linkedAdd/1_000_000, linkedGet/1_000_000, linkedIterate/1_000_000, linkedFront/1_000_000);
        System.out.printf("ArrayDeque: add/remove at both ends=%d ms%n", dequeEnds/1_000_000);
    }

    public static void main(String[] args) {
        test(1000);
        test(10000);
        test(100000);
    }
}
```

```python
%%writefile Attendance.java
import java.util.*;

class Student {
    private final String regNo;
    private final String name;

    Student(String regNo, String name) {
        this.regNo = regNo;
        this.name = name;
    }

    @Override
    public boolean equals(Object o) {
        if (this == o) return true;
        if (!(o instanceof Student)) return false;
        Student s = (Student) o;
        return regNo.equals(s.regNo);
    }

    @Override
    public int hashCode() {
        return Objects.hash(regNo);
    }

    @Override
    public String toString() {
        return regNo + " " + name;
    }
}

public class Attendance {
    public static void main(String[] args) {
        Set<Student> present = new HashSet<>();
        present.add(new Student("192311001", "Arun"));
        present.add(new Student("192311002", "Divya"));
        present.add(new Student("192311001", "Arun"));
        System.out.println("Unique students: " + present.size());

        // Additional test: same register number, different name.
        present.add(new Student("192311002", "Divya Kumar"));
        System.out.println("After duplicate scan: " + present.size());

        // TreeSet and HashMap alternatives.
        Set<String> treeSet = new TreeSet<>();
        treeSet.add("192311001");
        treeSet.add("192311002");
        treeSet.add("192311001");

        Map<String, String> byRegNo = new HashMap<>();
        byRegNo.put("192311001", "Arun");
        byRegNo.put("192311002", "Divya");
        System.out.println("TreeSet unique IDs: " + treeSet.size());
        System.out.println("HashMap unique IDs: " + byRegNo.size());
    }
}
```

```python
%%writefile PharmacyInventory.java
import java.util.*;

public class PharmacyInventory {
    static long lookupTime(Map<String, Integer> map, int lookups) {
        long start = System.nanoTime();
        long total = 0;
        for (int i = 0; i < lookups; i++) {
            total += map.getOrDefault("MED" + (i % map.size()), 0);
        }
        long elapsed = System.nanoTime() - start;
        if (total < 0) {
            System.out.println(total);
        }
        return elapsed;
    }

    static void run(int n) {
        Map<String, Integer> hash = new HashMap<>();
        Map<String, Integer> tree = new TreeMap<>();
        Map<String, Integer> linked = new LinkedHashMap<>();

        for (int i = 0; i < n; i++) {
            String code = String.format("MED%06d", i);
            hash.put(code, 100 + i);
            tree.put(code, 100 + i);
            linked.put(code, 100 + i);
        }

        System.out.println("\nMedicine count: " + n);
        System.out.printf("HashMap 100,000 lookups: %.3f ms%n", lookupTime(hash, 100000) / 1e6);
        System.out.printf("TreeMap 100,000 lookups: %.3f ms%n", lookupTime(tree, 100000) / 1e6);
        System.out.printf("LinkedHashMap 100,000 lookups: %.3f ms%n", lookupTime(linked, 100000) / 1e6);

        System.out.println("Code-order report (TreeMap): " +
                tree.keySet().stream().limit(5).toList());
        System.out.println("Insertion-order report (LinkedHashMap): " +
                linked.keySet().stream().limit(5).toList());
    }

    public static void main(String[] args) {
        run(100);
        run(1000);
        run(10000);
    }
}
```

```python
%%writefile LeaderboardDemo.java
import java.util.*;

class Team {
    final String name;
    final int points;
    final int goalDiff;

    Team(String name, int points, int goalDiff) {
        this.name = name;
        this.points = points;
        this.goalDiff = goalDiff;
    }

    @Override
    public String toString() {
        return name + "(" + points + "," + goalDiff + ")";
    }
}

class Leaderboard<T> {
    private final List<T> items = new ArrayList<>();
    private final Comparator<? super T> order;

    Leaderboard(Comparator<? super T> order) {
        this.order = order;
    }

    void add(T item) {
        items.add(item);
    }

    List<T> ranking() {
        List<T> copy = new ArrayList<>(items);
        copy.sort(order);
        return copy;
    }
}

public class LeaderboardDemo {
    public static void main(String[] args) {
        Comparator<Team> byRank =
                Comparator.comparingInt((Team t) -> t.points).reversed()
                        .thenComparing(Comparator.comparingInt((Team t) -> t.goalDiff).reversed())
                        .thenComparing(t -> t.name);

        Leaderboard<Team> board = new Leaderboard<>(byRank);
        board.add(new Team("Panthers", 9, 4));
        board.add(new Team("Eagles", 9, 6));
        board.add(new Team("Cobras", 6, 1));
        board.add(new Team("Bisons", 9, 4));

        System.out.println(board.ranking());
    }
}
```

```python
%%writefile Triage.java
import java.util.*;

class Patient {
    final String name;
    final int severity;
    final long arrival;

    Patient(String name, int severity, long arrival) {
        this.name = name;
        this.severity = severity;
        this.arrival = arrival;
    }

    @Override
    public String toString() {
        return name + "(S" + severity + ")";
    }
}

public class Triage {
    private long counter = 0;

    // Lower severity number means higher priority; earlier arrival wins ties.
    private final PriorityQueue<Patient> queue = new PriorityQueue<>(
            Comparator.comparingInt((Patient p) -> p.severity)
                    .thenComparingLong(p -> p.arrival));

    void admit(String name, int severity) {
        if (severity < 1 || severity > 5) throw new IllegalArgumentException("Severity must be 1 to 5");
        queue.add(new Patient(name, severity, counter++));
    }

    Patient next() {
        return queue.poll();
    }

    public static void main(String[] args) {
        Triage er = new Triage();
        er.admit("Ravi", 3);
        er.admit("Meena", 1);
        er.admit("John", 3);
        er.admit("Asha", 5);
        er.admit("Kumar", 1);
        er.admit("Lata", 2);

        StringBuilder order = new StringBuilder();
        Patient p;
        while ((p = er.next()) != null) {
            order.append(p).append(' ');
        }
        System.out.println(order.toString().trim());
    }
}
```

```python
%%writefile RedesignStringBuilder.java
public class RedesignStringBuilder {
    // Original approach: repeated String concatenation.
    static String original(int fields) {
        String ticket = "";
        for (int i = 1; i <= fields; i++) {
            ticket += "Field" + i + "=Value" + i + ";";
        }
        return ticket;
    }

    // Redesigned approach: append to one mutable StringBuilder.
    static String redesigned(int fields) {
        StringBuilder ticket = new StringBuilder();
        for (int i = 1; i <= fields; i++) {
            ticket.append("Field").append(i)
                  .append("=Value").append(i).append(';');
        }
        return ticket.toString();
    }

    static long measure(boolean useBuilder, int fields, int repeats) {
        long start = System.nanoTime();
        String result = "";
        for (int r = 0; r < repeats; r++) {
            result = useBuilder ? redesigned(fields) : original(fields);
        }
        long elapsed = System.nanoTime() - start;
        if (result.isEmpty()) System.out.println("Empty");
        return elapsed;
    }

    public static void main(String[] args) {
        int fields = 100;
        int repeats = 5000;

        // Warm-up
        measure(false, 20, 100);
        measure(true, 20, 100);

        long oldTime = measure(false, fields, repeats);
        long newTime = measure(true, fields, repeats);

        System.out.printf("Original String +: %.3f ms%n", oldTime / 1e6);
        System.out.printf("Redesigned StringBuilder: %.3f ms%n", newTime / 1e6);
        System.out.printf("Time ratio (original/new): %.2f%n", (double) oldTime / newTime);
        System.out.println("Both methods produce equivalent ticket text: " +
                original(fields).equals(redesigned(fields)));
    }
}
```
