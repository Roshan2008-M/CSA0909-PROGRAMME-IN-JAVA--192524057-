```python
# Clean up the extra comment/bash lines written to ATMWithdrawal.java by rewriting it correctly
%%writefile ATMWithdrawal.java
import java.util.*;

class InsufficientFundsException extends Exception {
    InsufficientFundsException(String message) {
        super(message);
    }
}

public class ATMWithdrawal {
    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);
        double balance = 5000;

        try {
            System.out.print("Enter Account Number: ");
            if (sc.hasNextInt()) {
                int account = sc.nextInt();
                System.out.print("Enter Withdrawal Amount: ");
                if (sc.hasNextDouble()) {
                    double amount = sc.nextDouble();

                    if (amount <= 0) {
                        throw new IllegalArgumentException("Invalid Amount");
                    }

                    if (amount > balance) {
                        throw new InsufficientFundsException("Insufficient Balance");
                    }

                    balance -= amount;
                    System.out.println("Withdrawal Successful");
                    System.out.println("Remaining Balance: " + balance);
                } else {
                    throw new InputMismatchException();
                }
            } else {
                throw new InputMismatchException();
            }
        } catch (InputMismatchException e) {
            System.out.println("Invalid Input");
        } catch (InsufficientFundsException e) {
            System.out.println(e.getMessage());
        } catch (IllegalArgumentException e) {
            System.out.println(e.getMessage());
        } finally {
            System.out.println("ATM Transaction Completed");
            sc.close();
        }
    }
}
```

```python
%%writefile StorageDemo.java
import java.io.IOException;

class ExamException extends Exception {
    ExamException(String message) {
        super(message);
    }

    ExamException(String message, Throwable cause) {
        super(message, cause);
    }
}

class SubmissionStorageException extends ExamException {
    SubmissionStorageException(String message, Throwable cause) {
        super(message, cause);
    }
}

class Storage {
    void save(String fileName, long sizeKb) throws SubmissionStorageException {
        try {
            if (sizeKb > 500) {
                throw new IOException("Disk quota exceeded");
            }
            System.out.println("Saved " + fileName);
        } catch (IOException e) {
            throw new SubmissionStorageException("Could not store " + fileName, e);
        }
    }
}

public class StorageDemo {
    public static void main(String[] args) {
        Storage storage = new Storage();

        try {
            storage.save("answer01.pdf", 120);
            storage.save("answer02.pdf", 800);
        } catch (ExamException e) {
            System.out.println("Error: " + e.getMessage());
            System.out.println("Cause: " + e.getCause().getMessage());
        }
    }
}
```

```python
# Compile the ScannerDemo program and filter out container warnings
!javac ScannerDemo.java 2>&1 | grep -v "warning" || true

# Run the ScannerDemo program and filter out container warnings
!java ScannerDemo 2>&1 | grep -v "warning"
```

```python
%%writefile SalesReportDynamic.java
import java.io.*;

public class SalesReportDynamic {
    public static void main(String[] args) {
        // Check if the user specified a filename via command-line arguments
        if (args.length < 1) {
            System.out.println("Usage: java SalesReportDynamic <input-filename>");
            return;
        }

        String inputFilename = args[0];

        try (BufferedReader br = new BufferedReader(new FileReader(inputFilename));
             BufferedWriter bw = new BufferedWriter(new FileWriter("summary.txt"))) {

            String line;
            while ((line = br.readLine()) != null) {
                bw.write(line);
                bw.newLine();
            }
            System.out.println("Sales Report Generated Successfully from: " + inputFilename);
        } catch (FileNotFoundException e) {
            System.out.println("File Error: The input file '" + inputFilename + "' was not found.");
        } catch (IOException e) {
            System.out.println("File Error: " + e.getMessage());
        }
    }
}
```

```python
# Compile the dynamic version
!javac SalesReportDynamic.java 2>&1 | grep -v "warning" || true

# Test 1: Run without providing any filename argument
print("--- Test 1: Running with no arguments ---")
!java SalesReportDynamic 2>&1 | grep -v "warning"

# Test 2: Run with the correct filename 'sales.txt' as an argument
print("\n--- Test 2: Running with 'sales.txt' argument ---")
!java SalesReportDynamic sales.txt 2>&1 | grep -v "warning"
```
